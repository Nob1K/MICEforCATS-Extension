```python
# Replace the string below with the path you just copied!
folder_path = "/content/drive/MyDrive/CSCI 5980 8980 Project/Notebooks/Data/combined_datasets_lite_hf"

# Load the dataset
combined_dataset = load_from_disk(folder_path)

# Convert to pandas
df_combined = combined_dataset.to_pandas()
print(df_combined.head())
```

```python
# 1. Get the unique question types (abstain, answer, clarify)
unique_types = df_combined['type'].unique()

df1_list = []
df2_list = []

# 2. Loop through each type to sample and split
for q_type in unique_types:
    # Filter the dataset for the current type
    type_subset = df_combined[df_combined['type'] == q_type]

    # Sample 332 unique rows (166 for dataset 1 + 166 for dataset 2)
    # Using random_state=8980 ensures reproducibility
    sampled_subset = type_subset.sample(n=332, random_state=8980)

    # Split the 332 rows into two mutually exclusive halves
    df1_list.append(sampled_subset.iloc[:166])
    df2_list.append(sampled_subset.iloc[166:])

# 3. Combine the lists into our final DataFrames
# We also use .sample(frac=1) to shuffle the final dataframes so the types aren't grouped together
df1 = pd.concat(df1_list).sample(frac=1, random_state=8980).reset_index(drop=True)
df2 = pd.concat(df2_list).sample(frac=1, random_state=8980).reset_index(drop=True)

# 4. Verify the output lengths and distributions
print("--- Dataset 1 ---")
print(f"Total shape: {df1.shape}")
print(df1['type'].value_counts())

print("\n--- Dataset 2 ---")
print(f"Total shape: {df2.shape}")
print(df2['type'].value_counts())

# 5. Export to CSV (Update the paths as needed for your Google Drive folder)
df1.to_csv('/content/drive/MyDrive/CSCI 5980 8980 Project/Notebooks/Data/dataset_lite_Brandon.csv', index=False)
df2.to_csv('/content/drive/MyDrive/CSCI 5980 8980 Project/Notebooks/Data/dataset_lite_Trek.csv', index=False)

print("\nSuccess! Both CSVs have been saved.")
```
