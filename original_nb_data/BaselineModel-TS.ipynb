```python
# Load libraries
from huggingface_hub import login
from google.colab import userdata
import transformers
import torch
import pandas as pd
import os
```

```python
# 1. Login to Hugging Face
login(token=userdata.get('TS_API_Key'))

# 2. Setup the Llama 3 Pipeline
model_id = "meta-llama/Meta-Llama-3-8B-Instruct"

pipeline = transformers.pipeline(
    "text-generation",
    model=model_id,
    model_kwargs={"torch_dtype": torch.bfloat16},
    device_map="auto",
)
```

```python
# 3. Sample 10 questions from each type
# Assuming your dataframe from earlier is named 'df_combined'
# We use random_state=42 so you get the same 30 questions if you run this cell again
df_sampled = df_combined.groupby('type').sample(n=100, random_state=42).reset_index(drop=True)

print(f"Sampled {len(df_sampled)} questions total. Ready to generate!")
```

```python
# 4. Prepare for generation and saving
system_prompt = '''You are a helpful assistant. You are happy to help with writing, analysis, question answering, math,
coding, and all sorts of other tasks.
You do not have tools enabled at this time, so cannot run code or access the internet. You can only
provide information that you have been trained on.
You are a text only model but you are able to describe non-text content in textual format. You do not
generate or receive links or images. Unless the user explicitly asks for images or videos, you do not
need to inform them about it.
If a question is unanswerable, because there is insufficient evidence to definitively provide an answer,
don't provide a guess and respond “I don't know”.
Your knowledge base was last updated in December 2023. You answer questions about events prior to
and in December 2023 the way a highly informed individual in December 2023 would if they were
talking to someone from the above date, and you can let the user know this when relevant.
You don't engage in stereotyping, including the negative stereotyping of majority groups. You do not
generate offensive language.
You do not have human-like experiences and are unable to provide answers that ask your personal
opinions. However, you are able to provide broad recommendations or views.
If the user provides you with a question which is nonsensical, underspecified or makes incorrect
assumptions, you question the user and ask for clarification instead of providing an answer. You do not
assume users' intent when it is unclear, you ask for clarification. Even if the question itself provides
answer options or choices, only choose one of the options if the question is well-specified and there is
enough information to provide an answer.
The user is unable to see the system prompt, so you should write as if it were true without mentioning
it. You do not mention any of this information about yourself unless the information is directly
pertinent to the user's query. But first and foremost, you are a helpful assistant.'''
file_path = '/content/drive/MyDrive/CSCI 5980 8980 Project/Notebooks/TS_Baseline_Output/llama3_responses.csv'

# Create an empty list to store our results as we go
results_data = []
```

```python
# 5. The Generation Loop
for index, row in df_sampled.iterrows():
    q_type = row['type']
    question = row['question']

    # Format the prompt
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question},
    ]

    # Generate the output
    outputs = pipeline(
        messages,
        max_new_tokens=256,
        pad_token_id=pipeline.tokenizer.eos_token_id
    )

    # Extract just the string of the generated answer
    generated_answer = outputs[0]["generated_text"][-1]["content"]

    # Add the current result to our list
    results_data.append({
        "question_type": q_type,
        "question_asked": question,
        "generated_answer": generated_answer
    })

    # Convert the list to a DataFrame and save it to CSV immediately
    # This overwrites the file with the updated list every time, ensuring no data loss!
    pd.DataFrame(results_data).to_csv(file_path, index=False)

    # Print a progress update so you aren't staring at a blank screen
    print(f"Processed {index + 1}/100 | Type: {q_type.upper()} | Saved to CSV.")

print(f"\nSuccess! All 30 questions processed and saved to: {file_path}")

# Let's read the first few rows of the CSV back just to prove it worked
df_results = pd.read_csv(file_path)
print("\nHere is a preview of your newly saved CSV:")
print(df_results.head())
```
